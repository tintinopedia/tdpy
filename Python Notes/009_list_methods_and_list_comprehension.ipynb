## List Methods

```python
# Append Method

list_items = [1, 2, 3, 4, 5, 6, 7, 8]
# append() operation would inser a value at the end of the list
list_items.append(9)
print(list_items)
```

```python
# insert()
list_items = [1, 2, 3, 4, 5, 6, 7, 8]
list_items.insert(2, "Avishek")
print(list_items)
```

```python
# remove()
list_items = [1, 2, 3, 4, 5, 6, 7, 8, 3]
list_items.remove(3)
print(list_items)
list_items.remove(3)
print(list_items)
```

```python
# pop()

list_items1 = [1, 2, 3, 4, 5, 6, 7, 8, 3]
list_items1.pop(2)
print(list_items)
```

```python
# index() 

list_items2 = [1, 2, 3, 4, 5, 6, 7, 8, 3]
index_value = list_items2.index(8)
print(index_value)
```

```python
# count()

list_items3 = [1, 8, 3, 2, 7, 4, 17, 6, 3]
list_items3.count(4)
print(list_items3)
list_items3.append(4)
print(list_items3.count(4))
```

```python
#sort()
 
list_items4 = [1, 8, 3, 2, 7, 4, 17, 6, 3]
list_items4.sort()
print(list_items4)
```

```python
# reverse sort

list_items4 = [1, 8, 3, 2, 7, 4, 17, 6, 3]
list_items4.reverse()
print(list_items4)
```

```python
# clear()

list_items5 = [1, 8, 3, 2, 7, 4, 17, 6, 3]
list_items5.clear()
print(list_items5)
```

## List Methods

#### append()--> insert the value at the end of the list   
#### insert()--->insert the value at a specific index location inside our list   
#### remove()--> remove the value from our list that we are going to pass inside it   
#### pop()---> remove the value from the list at a specific index location   
#### index()--> it provides the index of the value that we are going to pass inside it if that value is present inside that list   
#### sort()--> it is going to help us out with sorting the values in a ascending manner   
#### reverse()--> it can reverse the values present inside our list i.e, value at last index will be at first index now   
#### count()--> it counts the number of occurence of a specific value that we are going to pass inside our list and it returns that count of occurence   
#### clear()---> for removing all of the values that we are storing inside our list   

#### enumerate --  It helps us with getting the index value along with the actual value in a combination that we are storing inside the list.

```python
# enumerate()

Numbers = [ 4, 7, 2, 9, 4, 0, 1 ,5]
for index_value, value in enumerate(Numbers):
    print(index_value, value)
```

#### Length function = len() --> Gives us the length of the list that means that we will get the value of the total number of items we are storing inside the list.

```python
Numbers = [ 4, 7, 2, 9, 4, 0, 1 ,5]
length = len(Numbers)
print(length)
```

## List Comprehension

#### We are going to use Looping statement inside the list, so that we can store the values inside the list in an iterable format and in the list comprehension conditional statement are allowed as well.

#### Expressions for the item in iterable if condition

#### Expression --> what we want to print
#### for item in iterable --> looping statement
#### if condition --> conditional statement

```python
# Create a list storing the values from 0 to 10

# list_item6 = [0, 1, 4, 5, 6, 8, 10]
# print(list_item6)

list_item7 = [x for x in range(-1, 11)]
print(list_item7)
```

```python
list_squared = [y**3 for y in range(1, 11)]
print(list_squared)
```

```python
# If I am only interested in looking over the square values which are event

even_items = [x**2 for x in range(1,19) if x%2 == 0]
print(even_items)
```

```python
# Functions in list comprehension as well

list_item8 = ["Amit", "Apple", "Banana", "laptop", "technology"]
list_item9 = [len(word) for word in list_item8]
print(list_item9)
```

### Nested List Comprehension

#### We woould be having the combination of multiple lists in our single list itself. 

```python
# Nested List in comprehension

list_item10 = ["Amit", "Apple", "Banana", "laptop", "technology"]
list_item11 = ["Apple", "Banana", "Cider", "Maple", "Guava"]

pair_list = [[i, j] for i in list_item10 for j in list_item11]
print(pair_list)
```
