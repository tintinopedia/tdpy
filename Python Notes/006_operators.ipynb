## Types of Operators :
1. Arithmatic operators - Addition (+), Substraction (-), Product or Multiplication (*), Division (/), Modulus (%), Floor (//) and Exponent (a**b)
2. Comparison operators - Greater than (>), Lesser than (<), Greater than or equal to (<=), Lesser than or equal to (<=), Equal to (==) and Not equal to (!=)
3. Logical operators - and (whether both conditions are true or not), or (whether one of the operation is true or not), not (negation or to inverse the result of the given operation)
