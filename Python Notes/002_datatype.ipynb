## Data Types

## Data types are classifications of the data that tells the compiler or the interpreter, what kind of values we can store in to a variable and how much amount of memory will be consumed to store the data and the kind of operations it can perform.

```python
age1 = int(input("Enter the age:"))
age2 = int(input("Enter the age:"))
name = str(input("Enter your Name:"))
```

```python
print(age1)
print(age2)
```

```python
print(age1 + name)
```
