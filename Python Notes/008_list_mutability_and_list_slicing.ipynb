```python
lists = ["mango", "apple", "guava", "grapes", "pinapple", "banana"]
```

```python
print(lists[5])
```

```python
# Negative Indexing ----> In any data structure, we can access the data using Negative indexing as well.

print(lists[-1])
```

## Slicing

```python
# For having the access for all the values you are going to store inside the lists[], you can use these columns [:].

lists = ["mango", "apple", "guava", "grapes", "pinapple", "banana"]
print(lists[:])
```

## List Slicing

```python
# List slicing can help us out with retrieving a spcific data points from our list --> 

lists = ["mango", "apple", "guava", "grapes", "pinapple", "banana"]
lists[1:4]
```

```python
lists[1:3]
```

```python
# Print the above list in reverse order:

print(lists[::-1])
```

```python
veg = ["tomatoes", "brinjal", "potatoes"]
print(veg[0])
print(veg[-1])
print(veg[1])
print(veg[1:])
print(veg[0:4])
print(veg[::-1])
```

## List Mutability

#### By mentioning that Lists are mitable, we mean to say that we can replace the values present at a specific location inside our list. 

```python
lists[2]
```

```python
lists[2] = "Apple"
```

```python
print(lists[2])
```

```python
print(lists)
```

```python
# By using looping statements, we can iterate through List as well.

names = ["Avishek", "Ahaan", "Amit", "Swati", "Anamika", "Anil"]

for people in names:
    print(people)
```
